```python
from datetime import datetime
import random
from google import genai
from google.colab import userdata
!pip install edge-tts
import edge_tts
from IPython.display import Audio, display

# ── 1. 경기 결과 입력 (실습 1: 전반/후반 득점 분리) ──
player     = input("선수 이름은? ")
place      = input("경기가 열린 곳은? ")
opponent   = input("상대 팀은? ")
goals      = int(input("골 수는? "))
assists    = int(input("도움 수는? "))
score_us_1st   = int(input("우리 팀 전반 득점은? "))
score_them_1st = int(input("상대 팀 전반 득점은? "))
score_us_2nd   = int(input("우리 팀 후반 득점은? "))
score_them_2nd = int(input("상대 팀 후반 득점은? "))

score_us = score_us_1st + score_us_2nd
score_them = score_them_1st + score_them_2nd

# ── 2. 조사 고르기 ──
def josa(word, with_batchim, without_batchim):
    has_batchim = (ord(word[-1]) - 0xAC00) % 28 != 0
    return with_batchim if has_batchim else without_batchim

eul = josa(opponent, "을", "를")
gwa = josa(opponent, "과", "와")

# ── 3. 문장 재료 (실습 1: 각 항목당 문장 2개씩 추가) ──
OPENING = [
    "{player} 선수가 {place}에서 열린 {opponent}{gwa}의 경기에 선발 출전했습니다.",
    "{place}에서 펼쳐진 {opponent}{gwa}의 맞대결에 {player} 선수가 나섰습니다.",
    "수많은 팬들이 모인 {place}에서 {player} 선수가 {opponent}{gwa}의 경기에 출격했습니다.",
    "이날 {place}에는 {opponent}{gwa}의 치열한 승부를 위해 {player} 선수가 그라운드를 밟았습니다."
]

RESULT = {
    "승": [
        "최종 스코어 {us} 대 {them}. {opponent}{eul} 꺾고 승점 3을 챙겼습니다.",
        "치열한 공방 끝에 {us} 대 {them}으로 승리하며 {opponent}{eul} 완벽히 격파했습니다.",
        "종료 휘슬이 울릴 때 스코어는 {us} 대 {them}. 우리 팀이 {opponent}{eul} 상대로 값진 승리를 거두었습니다."
    ],
    "패": [
        "최종 스코어 {us} 대 {them}. {opponent}에게 아쉽게 무릎을 꿇었습니다.",
        "{us} 대 {them}으로 경기가 종료되며 {opponent}의 견고한 벽을 넘지 못했습니다.",
        "분전했으나 스코어 {us} 대 {them}으로 {opponent}에게 뼈아픈 패배를 당했습니다."
    ],
    "무": [
        "최종 스코어 {us} 대 {them}. 승부를 가리지 못했습니다.",
        "결국 {us} 대 {them}으로 비기며 {opponent}{gwa} 승점 1점씩을 나눠 가졌습니다.",
        "{us} 대 {them}의 팽팽한 흐름 속에 {opponent}{gwa}의 승부는 무승부로 마감되었습니다."
    ]
}
```

```python
PLAY = {
    (True,  True):  [
        "{goals}골 {assists}도움으로 경기를 지배했습니다.",
        "무려 {goals}골 {assists}도움을 기록하며 팀의 에이스다운 면모를 과시했습니다.",
        "공격의 선봉장으로서 {goals}득점 {assists}어시스트를 올리며 맹활약했습니다."
    ],
    (True,  False): [
        "{goals}골을 몰아치며 공격을 이끌었습니다.",
        "날카로운 득점 감각으로 {goals}골을 터뜨리며 팬들을 열광시켰습니다.",
        "상대 수비를 무너뜨리는 {goals}개의 득점으로 팀 공격의 핵심 역할을 수행했습니다."
    ],
    (False, True):  [
        "득점은 없었지만 {assists}개의 도움으로 힘을 보탰습니다.",
        "골맛은 보지 못했으나 완벽한 패스로 {assists}도움을 기록하며 결정적 찬스를 만들었습니다.",
        "이타적인 플레이가 빛났습니다. {assists}개의 어시스트로 공격의 든든한 윤활유가 되었습니다."
    ],
    (False, False): [
        "이날은 공격 포인트 없이 경기를 마쳤습니다.",
        "아쉽게도 골이나 도움을 기록하지는 못한 채 그라운드를 떠났습니다.",
        "활발한 움직임에도 불구하고 공격 포인트를 올리는 데는 아쉽게 실패했습니다."
    ]
}

# 실습 1: 전반/후반 득점 비교를 통한 경기 흐름 묘사
def get_flow_sentence():
    if score_us_1st < score_them_1st and score_us > score_them:
        return "특히 전반전의 열세를 뒤집고 짜릿한 역전승을 일구어냈습니다!"
    elif score_us_1st < score_them_1st and score_us == score_them:
        return "전반전 리드를 내줬으나 무서운 뒷심으로 추격하며 값진 무승부를 기록했습니다."
    elif score_us_1st > score_them_1st and score_us < score_them:
        return "아쉽게도 전반전의 리드를 지키지 못하고 역전패를 당하고 말았습니다."
    else:
        return ""

# ── 4. 판단 ──
def judge(us, them):
    if us > them: return "승"
    elif us < them: return "패"
    else: return "무"

# ── 5. 조립 ──
def write_article():
    now  = datetime.now().strftime("%Y-%m-%d %H:%M")
    head = f"[프리미어 리그 속보 · {now}]"
    body = [
        random.choice(OPENING),
        random.choice(RESULT[judge(score_us, score_them)]),
        get_flow_sentence(),
        random.choice(PLAY[(goals > 0, assists > 0)]),
    ]
    # 빈 문자열(흐름 문장이 없을 경우) 제외하고 결합
    body = [b for b in body if b]

    article = " ".join(body).format(
        player=player, place=place, opponent=opponent,
        goals=goals, assists=assists,
        us=score_us, them=score_them, eul=eul, gwa=gwa)
    return f"{head}\n{article}"

# LLM 다듬기 및 출력
raw_news = write_article()
client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))
prompt = ("아래 사실만 사용해서 자연스러운 5문장 기사로 다듬어 줘. "
          "사실을 새로 추가하지 말 것.\n" + raw_news)
answer = client.models.generate_content(model="gemini-3.6-flash", contents=prompt)
final_news = answer.text

print(final_news)
```

```python
# 실습 2: 남성 음성(ko-KR-InJoonNeural) 및 속도(+20%) 변경
VOICE = "ko-KR-InJoonNeural"
RATE = "+20%"

async def speak(text, filename="news.mp3", voice=VOICE, rate=RATE):
    tts = edge_tts.Communicate(text, voice=voice, rate=rate)
    await tts.save(filename)
    return filename

mp3 = await speak(final_news)
display(Audio(mp3, autoplay=True))
```
